# **Week 01: ERP Data Analysis 😁**

The objective of this analysis is to investigate the datasets available in the **company's ERP system**.The analysis will examine the structure and quality of the data, identify relationships between datasets, and perform an initial financial and sales analysis.

Task 1: Environment setup

```python
#import the required libraries
import pandas as pd # pandas: used to load and analyze datasets.
import numpy as np #numpy: used for numerical calculations.
import matplotlib.pyplot as plt #matplotlib.pyplot: used to create charts and visualizations.
```

Task 2: Load the datasets

```python
# Load datasets
base = "https://raw.githubusercontent.com/engrahmanimanel/-ERP-Data-Analysis/main/week01/" #Save the url common part in base.

customers = pd.read_csv(base + "customers.csv") #Combines the base URL with "customers.csv" and loads it into the customers DataFrame.
products = pd.read_csv(base + "products.csv")
sales = pd.read_csv(base + "sales.csv")
invoices = pd.read_csv(base + "invoices.csv")
payments = pd.read_csv(base + "payments.csv")

print("All datasets loaded successfully!")
```

```python
# Dataset overview
#This puts all five DataFrames into one dictionary.Makes it easier to analyze all datasets using one loop.

datasets = {
    "Customers": customers,
    "Products": products,
    "Sales": sales,
    "Invoices": invoices,
    "Payments": payments
}
for name, df in datasets.items():# This goes through each dataset one at a time. For example, the first time: name = "Customers" AND df = customers.
    print(f"\n{name}: {df.shape}") # df.shape shows (number of rows, number of columns).
    print("Columns:", list(df.columns)) # df.columns gets all column names then list them on a list.
    print("Missing:", df.isna().sum().sum()) #  isna(): checks for missing values.  First .sum(): counts missing values in each column and Second .sum(): adds all missing values together.
    print("Unique:", df.nunique().to_dict())# nunique() counts different values in each column.
    print("Types:\n", df.dtypes) #df.dtypes shows the data type of every column.
```

Task 3: Basic financial analysis

```python
# Financial Analysis(Number of customers,Total invoice value,Total amount paid)

print("Number of customers:", customers["customer_id"].nunique())
print("Total invoice value:", invoices["invoice_value"].sum())
print("Total amount paid:", payments["amount_paid"].sum())
```

Task 4: Initial Investigation

```python
# Product with highest revenue
product_revenue = sales.groupby("product_id")["revenue"].sum() # Groups all sales belonging to the same product together.
print("Highest-revenue product:", product_revenue.idxmax()) #id of maximum value.
print("Revenue:", product_revenue.max())
```

```python
# Customer with highest revenue
customer_revenue = (
    sales.merge(invoices[["invoice_id", "customer_id"]], on="invoice_id") # The sales dataset does not contain customer_id So we connect (merge) sales with invoices using their common column invoice_id.
         .groupby("customer_id")["revenue"]
         .sum()
)

print("Highest-revenue customer:", customer_revenue.idxmax())
print("Revenue:", customer_revenue.max())
```

```python
# Month with highest sales
sales["sale_date"] = pd.to_datetime(sales["sale_date"]) # pd.to_datetime(): Converts the text into a proper date format that pandas can analyze.

monthly_sales = sales.groupby(sales["sale_date"].dt.to_period("M"))["revenue"].sum() # This groups the sales by month and adds the revenue;.dt Allows us to work with parts of a date; "M" means month.

print("Highest-sales month:", monthly_sales.idxmax())
print("Sales:", monthly_sales.max())
```
